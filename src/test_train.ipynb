```python
from pathlib import Path
from ultralytics import YOLO
```

```python
dataset = "cctv_09062026_phone"
dataset_path = Path(f"datasets/{dataset}")
```

```python
model = YOLO("yolo11l.pt")

model.train(
    data=dataset_path,
    device=0,
    amp=True,
    epochs=100,
    patience=20,
    imgsz=640,
    batch=-1,
    workers=4,
    optimizer="SGD",
    lr0=0.01,
    cos_lr=True,
    mosaic=0.5,
    close_mosaic=15,
    scale=0.1,
    fliplr=0.5,
    hsv_h=0.02,
    hsv_s=0.25,
    hsv_v=0.15,
    cache=False,
    plots=True,
    save_period=10
)
```
