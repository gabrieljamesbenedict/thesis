```python
import pandas as pd
from pathlib import Path
import csv
```

```python
working_directory = Path("to_frames/")

for csv_file_path in working_directory.glob("*_true_summary.csv"):
    df = pd.read_csv(csv_file_path)
    
    frame_interactions = [["frame_index", "human_id", "object_id", "distance"]]
    for _, row in df.iterrows():
        for i in range(int(row["frame_start"]), int(row["frame_end"]), 5):
            frame_interactions.append([i, int(row["human_id"]), int(row["object_id"]), row["distance"]])
            
    output_name = working_directory / csv_file_path.name.replace("_true_summary.csv", "_true_frames.csv")
    with open(output_name, "w", newline="") as output_csv:
        csv.writer(output_csv).writerows(frame_interactions)
```
