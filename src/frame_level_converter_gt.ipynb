```python
import pandas as pd
from pathlib import Path
import csv
```

```python
output_directory = Path("video/gt/")

for csv_file_path in output_directory.glob("*_gt.csv"):

    csv_file = csv_file_path.stem
    csv_path = f"video/gt/{csv_file}.csv"
    print(csv_file)

    df = pd.read_csv(csv_path, header=None)

    frame_interactions = []
    for index, row in df.iterrows():
        start_frame = int(row[3])
        end_frame = int(row[4])
        h_id = int(row[0])
        o_id = int(row[1])
        for i in range(start_frame, end_frame):
            inters = [i, h_id, o_id]
            frame_interactions.append(inters)
            
    output_name = output_directory / (csv_file[:-3] + "_output.csv")
    with open(output_name, "w", newline="") as output_csv:
        csv.writer(output_csv).writerows(frame_interactions)
```
